```python
# Filter and select image
low_cloud_df = df[df['Cloud Cover'] < 30]
if low_cloud_df.empty:
    print("DataFrame is empty after filtering for Cloud Cover < 30.")
    raise ValueError("No image found with Cloud Cover < 30.")
sample_row = low_cloud_df.sort_values('Date').iloc[0]

# Fix path and load image
image_path_str = sample_row['image_path']
image_path_str = image_path_str.replace("raw_images/raw_images/", "raw_images/")
if image_path_str.startswith("../"):
    image_path_str = image_path_str[3:]
full_image_path = base_dir / image_path_str

print(f"\nLoading image: {sample_row['basename']}")
print(f"Date: {sample_row['Date']}")
print(f"Cloud Cover: {sample_row['Cloud Cover']}%")
print(f"Full path: {full_image_path}")
print(f"File exists: {full_image_path.exists()}")

try:
    with rasterio.open(full_image_path) as src:
        image_data = src.read()
        print(f"Successfully loaded image with shape: {image_data.shape}")
except rasterio.errors.RasterioIOError as e:
    print(f"\nError opening image at {full_image_path}")
    print(f"1. Check if parent directory exists: {full_image_path.parent.exists()}")
    if full_image_path.parent.exists():
        print(list(full_image_path.parent.glob('*'))[:10])
    print(f"{e}")
    raise
```

#### Prepare for ploting

```python
if image_data.shape[0] >= 3:
    image_plot = np.moveaxis(image_data[:3, :, :], 0, -1)
    plot_title = "RGB Image (Bands 1-3)"
    cmap = None
elif image_data.shape[0] == 1:
    image_plot = image_data[0, :, :]
    plot_title = "Single Band Image"
    cmap = 'gray'
else:
    raise ValueError(f"Unexpected image shape: {image_data.shape}")
```

#### Normalizing the image

```python
def normalize_percentile(band_data):
    p2 = np.percentile(band_data, 2)
    p98 = np.percentile(band_data, 98)
    scaled = (band_data - p2) / (p98 - p2 + 1e-9)
    return np.clip(scaled, 0, 1)
if image_plot.ndim == 3:
    plot_bands_normalized = []
    for i in range(image_plot.shape[2]):
        plot_bands_normalized.append(normalize_percentile(image_plot[:, :, i]))
    rgb_image_plot = np.stack(plot_bands_normalized, axis=-1)
else:
    rgb_image_plot = normalize_percentile(image_plot)
```

#### Ploting the graph

### Step 2: Feature Engineering

#### 1. Spectral Features (Simplified for RGB)

```python
from skimage import filters, feature, color
from scipy import ndimage
from scipy.stats import skew, kurtosis

print('Imported')
```

```python
def extract_spectral_features_rgb(image_data):

    features = {}
    
    red = image_data[0].astype(float)
    green = image_data[1].astype(float)
    blue = image_data[2].astype(float)
    
    # Band statistics
    features['red_mean'] = np.mean(red)
    features['green_mean'] = np.mean(green)
    features['blue_mean'] = np.mean(blue)
    
    features['red_std'] = np.std(red)
    features['green_std'] = np.std(green)
    features['blue_std'] = np.std(blue)
    
    # Color ratios (useful for distinguishing ice/snow from rock/water)
    features['red_green_ratio'] = np.mean(red) / (np.mean(green) + 1e-10)
    features['blue_green_ratio'] = np.mean(blue) / (np.mean(green) + 1e-10)
    features['red_blue_ratio'] = np.mean(red) / (np.mean(blue) + 1e-10)
    
    # Brightness (glaciers are typically bright)
    brightness = (red + green + blue) / 3
    features['brightness_mean'] = np.mean(brightness)
    features['brightness_std'] = np.std(brightness)
    features['brightness_max'] = np.max(brightness)
    
    # Saturation (glaciers have low saturation - grayish/white)
    max_rgb = np.maximum(np.maximum(red, green), blue)
    min_rgb = np.minimum(np.minimum(red, green), blue)
    saturation = (max_rgb - min_rgb) / (max_rgb + 1e-10)
    features['saturation_mean'] = np.mean(saturation)
    features['saturation_std'] = np.std(saturation)
    
    # Hue (color angle) - less useful for glaciers but can help identify vegetation
    features['hue_mean'] = np.mean(np.arctan2(np.sqrt(3)*(green-blue), 2*red-green-blue))
    
    return features
```

#### 2. Edge Features (Glacier Boundaries) using Sobel Operator


```python
def extract_edge_features_rgb(image_data):
    
    features = {}
    
    # Convert to grayscale
    gray = np.mean(image_data[:3], axis=0)
    
    # Sobel edge detection
    edges_sobel_h = filters.sobel_h(gray)
    edges_sobel_v = filters.sobel_v(gray)
    edges_sobel = np.sqrt(edges_sobel_h**2 + edges_sobel_v**2)
    
    features['edge_density'] = np.sum(edges_sobel > 0.1) / edges_sobel.size
    features['edge_mean'] = np.mean(edges_sobel)
    features['edge_std'] = np.std(edges_sobel)
    features['edge_max'] = np.max(edges_sobel)
    
    # Canny edges
    edges_canny = feature.canny(gray, sigma=2)
    features['canny_edge_count'] = np.sum(edges_canny)
    features['canny_edge_density'] = np.sum(edges_canny) / edges_canny.size
    
    # Terminus position (lowest edge point - glacier snout)
    edge_positions = np.where(edges_canny)
    if len(edge_positions[0]) > 0:
        features['terminus_y'] = np.max(edge_positions[0])  # Bottom-most edge
        features['terminus_x'] = np.mean(edge_positions[1][edge_positions[0] == features['terminus_y']])
        features['terminus_width'] = np.ptp(edge_positions[1])  # Width at terminus
        
        # Top of glacier
        features['glacier_top_y'] = np.min(edge_positions[0])
        features['glacier_length'] = features['terminus_y'] - features['glacier_top_y']
    else:
        features['terminus_y'] = 0
        features['terminus_x'] = 0
        features['terminus_width'] = 0
        features['glacier_top_y'] = 0
        features['glacier_length'] = 0
    
    return features
```

#### 3. Texture Features (Surface Characteristics)

```python
from skimage.feature import graycomatrix, graycoprops, local_binary_pattern

print('Imported')
```

```python
def extract_texture_features_rgb(image_data):
    """
    Texture analysis - glaciers have smooth ice vs rough rock
    """
    features = {}
    
    # Convert to grayscale
    gray = np.mean(image_data[:3], axis=0)
    
    # Normalize to 8-bit for GLCM
    gray_norm = ((gray - gray.min()) / (gray.max() - gray.min() + 1e-10) * 255).astype(np.uint8)
    
    # GLCM texture features
    distances = [1, 3, 5]
    angles = [0, np.pi/4, np.pi/2, 3*np.pi/4]
    
    glcm = graycomatrix(gray_norm, distances=distances, angles=angles,
                        levels=256, symmetric=True, normed=True)
    
    features['contrast'] = np.mean(graycoprops(glcm, 'contrast'))
    features['dissimilarity'] = np.mean(graycoprops(glcm, 'dissimilarity'))
    features['homogeneity'] = np.mean(graycoprops(glcm, 'homogeneity'))
    features['energy'] = np.mean(graycoprops(glcm, 'energy'))
    features['correlation'] = np.mean(graycoprops(glcm, 'correlation'))
    features['ASM'] = np.mean(graycoprops(glcm, 'ASM'))  # Angular Second Moment
    
    # Local Binary Pattern (LBP)
    radius = 3
    n_points = 8 * radius
    lbp = local_binary_pattern(gray_norm, n_points, radius, method='uniform')
    features['lbp_mean'] = np.mean(lbp)
    features['lbp_std'] = np.std(lbp)
    
    # Local standard deviation (roughness)
    features['local_std_3x3'] = np.mean(ndimage.generic_filter(gray, np.std, size=3))
    features['local_std_5x5'] = np.mean(ndimage.generic_filter(gray, np.std, size=5))
    features['local_std_7x7'] = np.mean(ndimage.generic_filter(gray, np.std, size=7))
    
    # Entropy (texture complexity)
    from skimage.filters.rank import entropy
    from skimage.morphology import disk
    try:
        entropy_img = entropy(gray_norm, disk(5))
        features['entropy_mean'] = np.mean(entropy_img)
        features['entropy_std'] = np.std(entropy_img)
    except:
        features['entropy_mean'] = 0
        features['entropy_std'] = 0
    
    return features
```

#### 4. Morphological Features (Shape & Area); Used Otsu Thresholding

```python
from skimage import morphology, measure
from skimage.filters import threshold_otsu

print('Imported')
```

```python
def extract_morphological_features_rgb(image_data):
    """
    Extract glacier shape and size
    """
    features = {}
    
    # Create binary mask (bright areas = glacier)
    gray = np.mean(image_data[:3], axis=0)
    
    # Otsu thresholding for automatic threshold
    try:
        thresh = threshold_otsu(gray)
        binary_mask = gray > thresh
    except:
        binary_mask = gray > 0.7  # Fallback threshold
    
    # Clean up mask
    binary_mask = morphology.remove_small_objects(binary_mask, min_size=50)
    binary_mask = morphology.remove_small_holes(binary_mask, area_threshold=50)
    
    # Basic area features
    features['glacier_area'] = np.sum(binary_mask)
    features['glacier_percentage'] = np.sum(binary_mask) / binary_mask.size * 100
    
    # Region properties
    labeled = measure.label(binary_mask)
    regions = measure.regionprops(labeled)
    
    if len(regions) > 0:
        # Get largest region (main glacier body)
        largest = max(regions, key=lambda x: x.area)
        
        features['perimeter'] = largest.perimeter
        features['eccentricity'] = largest.eccentricity  # Elongation
        features['solidity'] = largest.solidity  # Compactness
        features['extent'] = largest.extent
        features['major_axis'] = largest.major_axis_length
        features['minor_axis'] = largest.minor_axis_length
        features['orientation'] = largest.orientation  # Angle
        
        # Centroid (center of mass)
        features['centroid_y'] = largest.centroid[0]
        features['centroid_x'] = largest.centroid[1]
        
        # Bounding box
        bbox = largest.bbox
        features['bbox_height'] = bbox[2] - bbox[0]
        features['bbox_width'] = bbox[3] - bbox[1]
        features['bbox_area'] = features['bbox_height'] * features['bbox_width']
        
        # Convex hull (outline)
        features['convex_area'] = largest.convex_area
        features['filled_area'] = largest.filled_area
        
    else:
        # No glacier detected
        for key in ['perimeter', 'eccentricity', 'solidity', 'extent', 
                    'major_axis', 'minor_axis', 'orientation',
                    'centroid_y', 'centroid_x', 'bbox_height', 
                    'bbox_width', 'bbox_area', 'convex_area', 'filled_area']:
            features[key] = 0
    
    return features
```

#### 5. Statistical Features

```python
def extract_statistical_features_rgb(image_data):
    """
    Statistical measures for each band
    """
    features = {}
    
    band_names = ['red', 'green', 'blue']
    
    for i, name in enumerate(band_names):
        band = image_data[i].flatten()
        
        features[f'{name}_mean'] = np.mean(band)
        features[f'{name}_std'] = np.std(band)
        features[f'{name}_min'] = np.min(band)
        features[f'{name}_max'] = np.max(band)
        features[f'{name}_median'] = np.median(band)
        features[f'{name}_range'] = np.ptp(band)
        features[f'{name}_percentile_25'] = np.percentile(band, 25)
        features[f'{name}_percentile_75'] = np.percentile(band, 75)
        features[f'{name}_skewness'] = skew(band)
        features[f'{name}_kurtosis'] = kurtosis(band)
    
    return features
```

#### 6. Gradient Features

```python
def extract_gradient_features_rgb(image_data):
    """
    Analyze image gradients (rate of change)
    """
    features = {}
    
    gray = np.mean(image_data[:3], axis=0)
    
    # Gradients
    grad_y, grad_x = np.gradient(gray)
    grad_magnitude = np.sqrt(grad_x**2 + grad_y**2)
    
    features['gradient_mean'] = np.mean(grad_magnitude)
    features['gradient_std'] = np.std(grad_magnitude)
    features['gradient_max'] = np.max(grad_magnitude)
    
    # Gradient direction (flow direction)
    grad_direction = np.arctan2(grad_y, grad_x)
    features['gradient_direction_mean'] = np.mean(grad_direction)
    features['gradient_direction_std'] = np.std(grad_direction)
    
    return features
```

### Final Step: Complete Pipeline

```python
def extract_all_features_rgb(image_path, metadata_row):
    """
    Master function to extract ALL features from RGB image
    """
    # Load image
    with rasterio.open(image_path) as src:
        image_data = src.read()
    
    print(f"  Image shape: {image_data.shape}")
    
    # Initialize feature dictionary
    all_features = {}
    
    # Add metadata
    all_features['basename'] = metadata_row['basename']
    all_features['date'] = metadata_row['Date']
    all_features['cloud_cover'] = metadata_row['Cloud Cover']
    all_features['satellite'] = metadata_row['Satellite']
    all_features['region'] = metadata_row['Region']
    
    # Extract features
    print("  Extracting spectral features...")
    all_features.update(extract_spectral_features_rgb(image_data))
    
    print("  Extracting edge features...")
    all_features.update(extract_edge_features_rgb(image_data))
    
    print("  Extracting texture features...")
    all_features.update(extract_texture_features_rgb(image_data))
    
    print("  Extracting morphological features...")
    all_features.update(extract_morphological_features_rgb(image_data))
    
    print("  Extracting statistical features...")
    all_features.update(extract_statistical_features_rgb(image_data))
    
    print("  Extracting gradient features...")
    all_features.update(extract_gradient_features_rgb(image_data))
    
    return all_features

# Process all images
print("Starting feature extraction...")
feature_list = []

for idx, row in low_cloud_df.iterrows():
    print(f"\n[{idx+1}/{len(low_cloud_df)}] Processing: {row['basename']}")
    
    # Fix path
    image_path_str = row['image_path'].replace("raw_images/raw_images/", "raw_images/")
    if image_path_str.startswith("../"):
        image_path_str = image_path_str[3:]
    
    full_path = base_dir / image_path_str
    
    if full_path.exists():
        try:
            features = extract_all_features_rgb(full_path, row)
            feature_list.append(features)
            print("  ✓ Success")
        except Exception as e:
            print(f"  ✗ Error: {e}")
    else:
        print(f"  ✗ File not found: {full_path}")

# Convert to DataFrame
features_df = pd.DataFrame(feature_list)

# Save features
output_path = base_dir / 'glacier_features_rgb.csv'
features_df.to_csv(output_path, index=False)

print(f"\n{'='*60}")
print(f"Feature extraction complete!")
print(f"Total images processed: {len(features_df)}")
print(f"Total features extracted: {len(features_df.columns)}")
print(f"Saved to: {output_path}")
print(f"{'='*60}")

# Show feature summary
print("\nFeature categories:")
spectral_cols = [c for c in features_df.columns if any(x in c for x in ['ratio', 'brightness', 'saturation', 'hue'])]
edge_cols = [c for c in features_df.columns if 'edge' in c or 'terminus' in c or 'canny' in c]
texture_cols = [c for c in features_df.columns if any(x in c for x in ['contrast', 'homogeneity', 'lbp', 'entropy', 'local_std'])]
morph_cols = [c for c in features_df.columns if any(x in c for x in ['area', 'perimeter', 'centroid', 'bbox', 'axis'])]

print(f"  - Spectral: {len(spectral_cols)}")
print(f"  - Edge: {len(edge_cols)}")
print(f"  - Texture: {len(texture_cols)}")
print(f"  - Morphological: {len(morph_cols)}")

# Display first few rows
print("\nFirst few features:")
print(features_df.head())
```
